```python
import numpy as np
import time
from sklearn.linear_model import LogisticRegression
from sklearn.multioutput import MultiOutputClassifier

n_train, n_test, d_model = 5120, 1280, 576
X_train = np.random.randn(n_train, d_model).astype(np.float32)
X_test  = np.random.randn(n_test,  d_model).astype(np.float32)

y_train = np.random.randint(0, 2, (n_train, 9)).astype(int)
y_test  = np.random.randint(0, 2, (n_test,  9)).astype(int)

clf = MultiOutputClassifier(LogisticRegression(C=1.0, max_iter=1000))

t0 = time.perf_counter()
clf.fit(X_train, y_train)
t_fit = time.perf_counter() - t0
print(f"Temps du probe : {t_fit:.2f}s")
```
